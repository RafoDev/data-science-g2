# Normalización

## Tokenización

**cadena cruda/raw string**

```python
print(r"Hola amigos \nBuenas noches")
```

```python
texto = """Investigadora    \tde la Royal,Soctecy Dorochy
Hodgkin en el Instiruco de Neurociencia

Cognitiva del University College de Lon-
dres"""
print(texto)
```

### CASO 1: Tokenizar con espacios en blanco

```python
import re
corpus = re.split(r" ", texto)
print(corpus)
```

### CASO 2: Tokenizar con expresiones regulares

```python
corpus = re.split(r"[ \W\t\n]+", texto)
print(corpus)
```

## Tokenizar con NLTK



```python
texto = "En los E.U. esa postal vale $15.50. En Lon-dres vale $10.50 ..."
print(re.split(r"[ \W\t\n]+", texto))
```

```python
import nltk
```

```python
pattern = r'''(?x)     # set flag to allow verbose regexps
    (?:[A-Z]\.)+       # abbreviations, e.g. U.S.A.
  | \w+(?:-\w+)*       # words with optional internal hyphens
  | \$?\d+(?:\.\d+)?%? # currency and percentages, e.g. $12.40, 82%
  | \.\.\.             # ellipsis
  | [][.,;"'?():-_`]   # these are separate tokens; includes ], [
'''
```

```python
corpus = nltk.regexp_tokenize(texto, pattern)
corpus
```
