```python
import nltk
nltk.download('book')
from nltk.book import *
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

# Bi-grama

```python
bigramas = list(bigrams(text1))
```

```python
fdist = FreqDist(bigramas)
list(fdist.items())[:10]
```

```python
fdist.most_common(10)
```

```python
fdist.plot(10)
```

## Filtrado de bigramas

```python
min_letras = 5
bigramas_filtrado = [b for b in bigramas if len(b[0]) > min_letras and len(b[1]) > min_letras ]
bigramas_filtrado
```

```python
fdist_filtrado = FreqDist(bigramas_filtrado)
fdist_filtrado.plot(10)
```

# Trigramas

```python
from nltk.util import ngrams
trigramas = list(ngrams(text1, 3))
trigramas[:10]
```

```python
min_letras = 2
trigramas_filtrado = [b for b in trigramas if len(b[0]) > min_letras and len(b[1]) > min_letras and len(b[2]) > min_letras ]
trigramas_filtrado
```

```python
fdist_filtrado = FreqDist(trigramas_filtrado)
fdist_filtrado.most_common(10)
```

```python
fdist_filtrado.plot(10)
```
