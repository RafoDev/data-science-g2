# Numerizar la data

# Crear DataLoaders

```python
def get_collate_fn(pad_index):
  def collate_fn(batch):
    # 1. Procesamiento de IDs
    batch_ids = [i["ids"] for i in batch]
    # 2. Padding de Secuencias
    batch_ids = nn.utils.rnn.pad_sequence(
        batch_ids,
        padding_value=pad_index,
        batch_first=True
    )
    # 3. Procesamiento de etiquetas
    batch_labels = [i["label"] for i in batch]
    batch_labels = torch.stack(batch_labels)

    # 4. Crear batch final
    batch = {
        "ids": batch_ids,
        "labels": batch_labels
    }

    return batch
  return collate_fn
```

```python
def get_data_loader(dataset, batch_size, pad_index, shuffle=False):
  collate_fn = get_collate_fn(pad_index)
  data_loader = torch.utils.data.DataLoader(
      dataset=dataset,
      batch_size=batch_size,
      collate_fn=collate_fn,
      shuffle=shuffle
  )
  return data_loader
```

```python
BATCH_SIZE = 256

train_data_loader = get_data_loader(train_data, BATCH_SIZE, pad_index, shuffle=True)
val_data_loader = get_data_loader(val_data, BATCH_SIZE, pad_index, shuffle=False)
test_data_loader = get_data_loader(test_data, BATCH_SIZE, pad_index, shuffle=False)
```

# Crear Modelo de Neural Bag Of Words

```python
class NBow(nn.Module):
  def __init__(self, vocab_size, embedding_dim, output_dim, pad_index):
    super().__init__()
    self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=pad_index)
    self.fc = nn.Linear(embedding_dim, output_dim)

  def forward(self, ids):
    embedded = self.embedding(ids)
    pooled = embedded.mean(dim=1)
    prediction = self.fc(pooled)
    return prediction
```

```python
vocab_size=len(vocab)
embedding_dim = 300
output_dim = len(train_data.unique("label"))

model = NBow(vocab_size, embedding_dim, output_dim, pad_index)
```

# Compilación del modelo

# Entrenamiento del modelo

```python
import numpy as np
import collections
```

```python
def get_accuracy(prediction, labels):
  batch_size,_ = prediction.shape
  predicted_classes = prediction.argmax(dim=-1)
  correct_predictions = predicted_classes.eq(labels).sum()
  accuracy = correct_predictions / batch_size
  return accuracy
```

```python
def train(data_loader, model, criterion, optimizador, device):
  model.train()
  epoch_losses = []
  epoch_accs = []

  for batch in tqdm.tqdm(data_loader, desc="training..."):
    ids = batch["ids"].to(device)
    labels = batch["labels"].to(device)

    prediction = model(ids)
    loss = criterion(prediction, labels)
    accuracy = get_accuracy(prediction, labels)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    epoch_losses.append(loss.item())
    epoch_accs.append(accuracy.item())

  return np.mean(epoch_losses), np.mean(epoch_accs)
```

```python
def evaluate(data_loader, model, criterion, device):
  model.eval()
  epoch_losses = []
  epoch_accs = []

  with torch.no_grad():
    for batch in tqdm.tqdm(data_loader, desc="evaluating..."):
      ids = batch["ids"].to(device)
      labels = batch["labels"].to(device)

      prediction = model(ids)
      loss = criterion(prediction, labels)
      accuracy = get_accuracy(prediction, labels)

      epoch_losses.append(loss.item())
      epoch_accs.append(accuracy.item())
  return np.mean(epoch_losses), np.mean(epoch_accs)
```

```python
n_epochs = 10
best_valid_loss = float("inf")
metrics = collections.defaultdict(list)

for epoch in range(n_epochs):
  train_loss, train_acc = train(train_data_loader, model, criterion, optimizer, device)
  val_loss, val_acc = evaluate(val_data_loader, model, criterion, device)

  metrics["train_losses"].append(train_loss)
  metrics["train_accs"].append(train_acc)
  metrics["val_losses"].append(val_loss)
  metrics["val_accs"].append(val_acc)

  if val_loss < best_valid_loss:
    best_valid_loss = val_loss
    torch.save(model.state_dict(), "nbow.pt")

  print(f"epoch: {epoch}")
  print(f"train_loss: {train_loss:.3f}, train_acc: {train_acc:.3f}, ")
  print(f"val_loss: {val_loss:.3f}, val_acc: {val_acc:.3f}, ")
```

# Evaluar con data

```python
def predict_sentiment(text, model, tokenizer, vocab, device):
  tokens = tokenizer(text)
  ids = vocab.lookup_indices(tokens)
  tensor = torch.LongTensor(ids).unsqueeze(dim=0).to(device)
  prediction = model(tensor).squeeze(dim=0)
  probability = torch.softmax(prediction, dim=-1)
  predicted_class = prediction.argmax(dim=-1).item()
  predicted_probability = probability[predicted_class].item()
  return predicted_class, predicted_probability
```

```python
text = "this movie is great!"
predict_sentiment(text, model, tokenizer, vocab, device)
```

```python
text = "this movie is terrible!"
predict_sentiment(text, model, tokenizer, vocab, device)
```

```python
text = "this movie is good, but can be better"
predict_sentiment(text, model, tokenizer, vocab, device)
```
