## Calcular la Riqueza Léxica

Es una medida de la variedad y diversidad del vocabulario utilizado en un texto.

```python
# palabras únicas / total de palabras = riqueza léxica
vocabulario = sorted(set(text1))
```

```python
rl = len(vocabulario)/len(text1)
rl
```

```python
def porcentaje_palabra(palabra, texto):
  return text1.count(palabra)/len(texto) * 100
```

```python
porcentaje_palabra("a", text2)
```

# Distribuciones

```python
dic_palabras = {}
for palabra in vocabulario: # Recorremos la lista sin duplicados
  entry = {palabra: text1.count(palabra)} # Contamos en la lista original
  dic_palabras.update(entry)

dic_palabras
```

```python
dic_ordenado = dict(sorted(dic_palabras.items(), key=lambda item: item[1], reverse=True))
dic_ordenado
```

```python
top_10 = list(dic_ordenado.items())[:10]
top_10
```

```python
tokens = [item[0] for item in top_10]
frecuencias = [item[1] for item in top_10]

plt.bar(tokens, frecuencias)
plt.xlabel("Tokens")
plt.ylabel("Freq")
plt.show()
```

# Filtrado utilizando stopwords

**stopwords**: son palabras comunes que no aportan significado relevante. Ej: el, la, de, que, un

```python
nltk.download("stopwords")
from nltk.corpus import stopwords
```

```python
stop_words = stopwords.words("english")
stop_words
```

```python
words = [word for word in text1 if word.isalpha()]
words[:10]
```

```python
filtered_words = [word for word in words if word not in stop_words]
filtered_words[:10]
```

```python
fdist_filtered = FreqDist(filtered_words)
fdist_filtered.most_common(10)
```

```python
fdist_filtered.plot(10)
```
