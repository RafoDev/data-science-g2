```python
import nltk
nltk.download("treebank")
from nltk.corpus import treebank
```

```python
train_data = treebank.tagged_sents()[:3900]
train_data
```

```python
from nltk.tag import hmm
tagger = hmm.HiddenMarkovModelTrainer().train_supervised(train_data, estimator=lambda fd, bins: nltk.probability.LidstoneProbDist(fd, 0.1, bins))
tagger
```

```python
tagger.tag("Pierre Vinken, 61 years old".split())
```

```python
tagger.accuracy(treebank.tagged_sents()[3901:])
```
