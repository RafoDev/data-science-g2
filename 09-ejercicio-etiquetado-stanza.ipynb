```python
import pandas as pd

df = pd.read_csv("/content/vacunasXFrase.csv", sep="|")
df.head(5)
```

```python
df_tweets = df["tweet"].head(10)
df_tweets
```

```python
import stanza
stanza.download("es")
```

```python
# https://stanfordnlp.github.io/stanza/pipeline.html#processors
nlp = stanza.Pipeline("es")
```

```python
def tag_tweet(tweet):
  doc = nlp(tweet)
  tags = [(word.text, word.upos) for sent in doc.sentences for word in sent.words]
  return tags
```

```python
tags = tag_tweet("@Patomorenoc Hace algunos años había una serie")
tags
```

```python
df_tags = df_tweets.apply(tag_tweet)
```
