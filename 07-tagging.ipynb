# Tagging en Español

```python
nltk.download("cess_esp")
from nltk.corpus import cess_esp as cess
from nltk import UnigramTagger
```

### Entrenamiento de UnigramTagger

```python
cess_sents = cess.tagged_sents()
```

```python
fraction = int(len(cess_sents)*90/100)
fraction
```

```python
uni_tagger = UnigramTagger(cess_sents[:fraction])
```

```python
uni_tagger.accuracy(cess_sents[fraction+1:])
```

```python
uni_tagger.tag("Hoy trabajé en la tarea ds-233".split(" "))
```

### Entrenamiento con BigramTagger

```python
from nltk import BigramTagger
```

```python
bi_tagger = BigramTagger(cess_sents[:fraction])
bi_tagger.accuracy(cess_sents[fraction+1:])
bi_tagger.tag("Hoy trabajé en la tarea ds-233".split(" "))
```
